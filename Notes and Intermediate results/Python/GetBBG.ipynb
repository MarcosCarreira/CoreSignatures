### Get Historical Data - BMF - DI1

```python
DI_fields = ['PX_SETTLE', 'EXCH_YEST_ADJ_ALT_SETT_RATE', 'FUT_BUS_DAYS_VAL', 'VOLUME', 'RT_OPEN_INTEREST']
```

```python
'ODF24 Comdty'.split()[0]
```

```python
def get_future(ticker, start, today):
    query_dates = blp.bdp(tickers=ticker, flds=['FUT_FIRST_TRADE_DT', 'FUT_DLV_DT_FIRST'])
    start_dt = max(start, query_dates['fut_first_trade_dt'].values[0])
    end_dt = min(today, query_dates['fut_dlv_dt_first'].values[0])
    df = blp.bdh(tickers=ticker, flds=DI_fields,
       start_date=start_dt, end_date=end_dt)
    df_cols = df.columns.values
    df.columns = [c[1] for c in df_cols]
    df['CONTRACT'] = ticker.split()[0]
    filename = ticker + '.csv'
    df.to_csv(w_path_BMF + filename)
    print(filename + ' saved')
```

```python
for t in all_tickers:
    get_future(t, start, today)
```

```python
df_all_DI1 = pd.DataFrame()
for t in all_tickers:
    df = pd.read_csv(w_path_BMF + t + '.csv', index_col=0)
    cols = ['Date'] + list(df.columns)
    df = df.reset_index()
    df.columns = cols
    df['MATURITY'] = df['FUT_BUS_DAYS_VAL'] / 252
    df_all_DI1 = pd.concat([df_all_DI1, df])
df_all_DI1  = df_all_DI1.set_index(['Date', 'MATURITY'])
```

### Get Historical Data - SOFR - Rates

```python
def get_SOFR(ticker, start, today):
    df = blp.bdh(tickers=ticker, flds=['PX_LAST'],
       start_date=start, end_date=today)
    df_cols = df.columns.values
    df.columns = [c[1] for c in df_cols]
    df['MATURITY'] = ticker.split()[0][6:]
    filename = ticker + '.csv'
    df.to_csv(w_path_SOFR + filename)
    print(filename + ' saved')
```
